## 5. Documentation

- **Environment creation command:** `python -m venv env` (or `conda create -n fe-course python=3.11 -y`), then `pip install python-dotenv numpy jupyter`.
- **Why `.env` matters:** it keeps secrets like API keys and data paths out of source control. `.env.example` is committed as a template with dummy values, `.env` itself is gitignored and holds the real values, and `src/config.py` reads through it so no notebook ever hardcodes a key.
- **What the NumPy demo shows:** a vectorized operation (`arr * 10`) applied to the whole array at once instead of looping in Python. It's the pattern every later stage's numeric code relies on.
